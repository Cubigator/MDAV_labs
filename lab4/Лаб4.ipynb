3.  **Предсказание исходов игры (классификация):**
    *   Прогнозирование победителя на основе агрегированных продвинутых метрик, а не только счетов прошлых игр.

    *   Анализ времени на льду (`shifts`) для поиска оптимальных сочетаний игроков.

#### Не подходит для (или подходит с оговорками):

1.  **Простого анализа "из коробки":** Этот датасет категорически не подходит для быстрого построения модели в `scikit-learn` без предварительной многочасовой инженерной работы по объединению таблиц, очистке и агрегации событий. Требуются навыки SQL или продвинутого `pandas`.
2.  **Моделирования травм и физического состояния:** Хотя есть данные о силовых приемах (`hits`), нет прямых медицинских показателей, данных об усталости по скорости движения или биометрических данных.
3.  **Прогнозирования с учетом рыночных коэффициентов:** В датасете нет букмекерских линий. Целевую переменную "победитель" нужно сравнивать со скрытыми вероятностями рынка для поиска "валуйных" ставок, но самих вероятностей здесь нет.
4.  **Анализа драфта и развития юниоров:** Данные строго об играх НХЛ, нет статистики выступлений в низших лигах (AHL, юниорские лиги), которая используется для скаутинга.

Я разделил все гипотезы на 3 уровня:
* 1 уровень - Базовые закономерности игры
* 2 уровень - Продвинутая аналитика и игровые события
* 3 уровень - Игроки и микроматчи

На каждый уровень я выбрал несколько гипотез:

**Гипотезы 1 уровня:**
1) Преимущество своего льда: Команды-хозяева выигрывают значимо чаще, чем гости
2) Падение качества к концу матча: Процент реализации бросков в третьем периоде ниже, чем в первом
3) Финал сезона — более оборонительный хоккей: В последние 10 игр регулярного чемпионата средняя результативность ниже, чем в первые 10 игр
4) Давление трибун на судей: Хозяева получают в среднем меньше минут штрафа за матч, чем гости
5) Плей-офф — меньше голов: Средний тотал матча в плей-офф ниже, чем в регулярном чемпионате
6) Октябрь — месяц вратарей: В октябре (начало сезона) забивается меньше всего голов по сравнению с другими месяцами

**Гипотезы 2 уровня:**
1) Дальние броски бесполезны: Броски с расстояния более 50 футов имеют вероятность гола менее 2%
2) Тип броска решает: Броски "в касание" (one-timer) имеют более высокий процент реализации, чем броски с обычного замаха (wrist shot)
3) Реализация большинства: Команда в большинстве забивает гол в первые 30 секунд реализации численного преимущества чаще, чем в оставшееся время
4) Волевые победы редки: Команда, проигрывающая после 2-го периода, имеет менее 20% шансов на итоговую победу в основное время
5) Первый гол решает: Команда, забившая первой, выигрывает матч более чем в 65% случаев

**Гипотезы 3 уровня:**
1) Пик результативности: Пик средней результативности нападающих приходится на возраст 27-28 лет
2) Защитники "созревают" позже: Защитники достигают пика результативности в среднем на 2 года позже нападающих (получается 29-30 лет)

4) Европейцы против североамериканцев: Европейские защитники делают в среднем меньше силовых приемов (hits), но больше перехватов, чем североамериканские

```python
def transform_outcome(result: str):
    if 'home' in result.lower():
        return 'Хозяева'
    elif 'away' in result.lower():
        return 'Гости'
    return 'NaN'

win_data = game_data['outcome'].apply(transform_outcome)
```

### 2 Гипотеза : Падение качества к концу матча: Процент реализации бросков в третьем периоде ниже, чем в первом

```python
game_plays_data = pd.read_csv('/content/drive/MyDrive/archive/game_plays.csv')
```

```python
shots_data = game_plays_data[(game_plays_data['event'].isin(['Shot', 'Goal'])) & (game_plays_data['period'] <= 3)]
```

```python
def calculate_conversion_rate(data):
    total_goals = len(data[data['event'] == 'Goal'])
    total_shots = len(data[data['event'] == 'Shot']) + total_goals
    return total_goals / total_shots * 100
```

```python
period_stats = shots_data.groupby('period').apply(calculate_conversion_rate).reset_index()
period_stats.columns = ['period', 'conversion_rate']
```

```python
g = sns.catplot(
    data=period_stats, kind="bar",
    x="period", y="conversion_rate",
    errorbar="sd", palette="dark", height=6
)

g.set_axis_labels("Период", "Реализация (%)")
g.fig.suptitle("Процент реализации бросков по периодам", y=1.02)

ax = g.ax

for i, p in enumerate(ax.patches):
    height = p.get_height()
    if not np.isnan(height):
        ax.text(
            p.get_x() + p.get_width()/2.,
            height + 0.5,
            f'{height:.1f}%',
            ha='center',
            va='bottom',
            fontsize=10,
            fontweight='bold'
        )

plt.tight_layout()
plt.show()
```

Вместо ожидаемого падения наблюдается устойчивый рост реализации от первого периода к третьему.

**Вывод:** Гипотеза **полностью опровергнута**. Обнаружена обратная закономерность: реализация бросков не падает, а, наоборот, растет к концу матча. Возможные объяснения: усталость вратарей, накопленный темп игры, более рискованная тактика проигрывающих команд в третьем периоде, либо оголение тылов при попытках отыграться. Это интересное наблюдение, которое стоит учесть в дальнейшем анализе — возможно, усталость влияет не на качество бросков, а на способность вратарей их отражать.

### 3 гипотеза : Финал сезона — более оборонительный хоккей: В последние 10 игр регулярного чемпионата средняя результативность ниже, чем в первые 10 игр

```python
game_data = pd.read_csv('/content/drive/MyDrive/archive/game.csv')
game_data = game_data[game_data['type'] == 'R']
game_data = game_data[(game_data['season'] != 20182019) & (game_data['season'] != 20192020) & (game_data['season'] != 20122013)]
```

```python
game_data = game_data.sort_values('date_time_GMT')
game_data['total_goals'] = game_data['home_goals'] + game_data['away_goals']
game_data['home_team_game_number'] = 0
game_data['away_team_game_number'] = 0
```

```python
buffer = dict()

for idx, row in game_data.iterrows():
    key_1 = str(row['season']) + ':' + str(row['home_team_id'])
    key_2 = str(row['season']) + ':' + str(row['away_team_id'])
    buffer[key_1] = buffer.get(key_1, 0) + 1
    buffer[key_2] = buffer.get(key_2, 0) + 1
    game_data.at[idx, 'home_team_game_number'] = buffer[key_1]
    game_data.at[idx, 'away_team_game_number'] = buffer[key_2]
```

```python
game_data_start_season = game_data[(game_data['home_team_game_number']) <= 10 | (game_data['away_team_game_number'] <= 10)]
game_data_end_season = game_data[(game_data['home_team_game_number']) >= 72 | (game_data['away_team_game_number'] >= 72)]
```

```python
seasons_data = []

seasons = set(game_data_start_season['season'])

for season_num in seasons:
    avg_start = game_data_start_season[game_data_start_season['season'] == season_num]['total_goals'].mean()
    avg_end = game_data_end_season[game_data_end_season['season'] == season_num]['total_goals'].mean()

    seasons_data.append({
        'season': season_num,
        'period': 'Начало сезона',
        'avg_goals': avg_start
    })
    seasons_data.append({
        'season': season_num,
        'period': 'Конец сезона',
        'avg_goals': avg_end
    })

comparison_df = pd.DataFrame(seasons_data)

plt.figure(figsize=(12, 6))
sns.barplot(data=comparison_df,
            x='season',
            y='avg_goals',
            hue='period',
            palette={'Начало сезона': 'skyblue', 'Конец сезона': 'lightcoral'})

plt.title('Сравнение средней результативности\nНачало vs Конец сезона',
          fontsize=14, fontweight='bold')
plt.xlabel('Сезон', fontsize=12)
plt.ylabel('Среднее количество голов за матч', fontsize=12)
plt.legend(title='Период')
plt.xticks(rotation=45)

for i, p in enumerate(plt.gca().patches):
    height = p.get_height()
    plt.gca().text(p.get_x() + p.get_width()/2., height + 0.05,
                   f'{height:.2f}', ha="center", fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

**Вывод:** Гипотеза **частично подтвердилась**. Однако эффект не универсален — в 25% сезонов результативность, наоборот, росла. Это может объясняться особенностями конкретных сезонов: травмами ключевых вратарей, изменениями правил или высокой плотностью в турнирной таблице, заставляющей команды рисковать

### 4 гипотеза : Давление трибун на судей: Хозяева получают в среднем меньше минут штрафа за матч, чем гости

```python
# game_plays_data = pd.read_csv('/content/drive/MyDrive/archive/game_plays.csv')
game_data = pd.read_csv('/content/drive/MyDrive/archive/game.csv')
game_data = game_data[(game_data['season'] != 20182019) & (game_data['season'] != 20192020) & (game_data['season'] != 20122013)]
game_penalties = pd.read_csv('/content/drive/MyDrive/archive/game_penalties.csv')
penalty_data = game_plays_data[game_plays_data['event'] == 'Penalty']
penalty_data = penalty_data.merge(
    game_data[['game_id', 'home_team_id', 'away_team_id']],
    on='game_id',
    how='left'
)
penalty_data = penalty_data.merge(
    game_penalties[['play_id', 'penaltyMinutes']],
    on='play_id',
    how='left'
)

game_data['home_total_penalty'] = 0
game_data['away_total_penalty'] = 0

for idx, row in game_data.iterrows():
    home_total_penalty = penalty_data[(penalty_data['game_id'] == row['game_id']) & (penalty_data['team_id_against'] == penalty_data['home_team_id'])]['penaltyMinutes'].sum()
    away_total_penalty = penalty_data[(penalty_data['game_id'] == row['game_id']) & (penalty_data['team_id_against'] == penalty_data['away_team_id'])]['penaltyMinutes'].sum()
    game_data.at[idx, 'home_total_penalty'] = home_total_penalty
    game_data.at[idx, 'away_total_penalty'] = away_total_penalty
```

```python
game_data_melted = game_data.melt(
    id_vars=['season'],
    value_vars=['home_total_penalty', 'away_total_penalty'],
    var_name='team_type',
    value_name='penalty_minutes'
)

game_data_melted['team_type'] = game_data_melted['team_type'].replace({
    'home_total_penalty': 'home',
    'away_total_penalty': 'away'
})

plt.figure(figsize=(10, 6))

sns.boxplot(data=game_data_melted,
            x='team_type',
            y='penalty_minutes',
            palette={'home': 'lightblue', 'away': 'lightcoral'})

plt.yscale('log')

plt.title('Boxplot (с логарифмической шкалой)',
          fontsize=14, fontweight='bold')
plt.xlabel('Тип команды', fontsize=12)
plt.ylabel('Штрафные минуты (логарифмическая шкала)', fontsize=12)
plt.xticks([0, 1], ['Дома', 'В гостях'])
plt.show();
```

**Вывод:** Гипотеза **опровергнута**. Вопреки распространенному мнению о давлении трибун на судей, данные НХЛ не показывают статистически значимой разницы в штрафных минутах между хозяевами и гостями. Это можно интерпретировать двояко: либо судьи НХЛ достаточно профессиональны, чтобы не поддаваться влиянию болельщиков, либо давление трибун компенсируется другими факторами (например, гости чаще фолят из-за более агрессивной игры хозяев или усталости после перелетов)

### Гипотеза 7 : Дальние броски бесполезны: Броски с расстояния более 50 футов имеют вероятность гола менее 2%

```python
shots_data = game_plays_data[(game_plays_data['event'].isin(['Shot', 'Goal']))]
```

```python
first_period = shots_data[shots_data['period'] == 1]

# Группируем по игре и команде, берем первый бросок/гол в 1-м периоде
first_shot_per_team = first_period.groupby(['game_id', 'team_id_for']).first().reset_index()

# Создаем словарь: (game_id, team_id_for) -> goal_x для нечетных периодов
team_direction_dict = {}
for _, row in first_shot_per_team.iterrows():
    if pd.notna(row['x']):
        if row['x'] < 0:
            goal_x = -89.0  # Атакуют ворота слева
        else:
            goal_x = 89.0   # Атакуют ворота справа
        team_direction_dict[(row['game_id'], row['team_id_for'])] = goal_x

def get_goal_x(row):
    key = (row['game_id'], row['team_id_for'])
    base_goal = team_direction_dict.get(key, np.nan)

    if pd.isna(base_goal):
        return np.nan

    # В чётных периодах (2-й период) команды меняются воротами
    if row['period'] == 2:
        return -base_goal
    else:  # 1-й, 3-й период, овертайм
        return base_goal

shots_data.loc[:, 'goal_x'] = shots_data.apply(get_goal_x, axis=1)
shots_data.loc[:, 'goal_y'] = 0.0  # Y-координата ворот всегда 0
```

```python
def get_distance(data):
    return np.sqrt((data['x'] - data['goal_x']) ** 2 + data['y'] ** 2)


def calculate_conversion_rate(data):
    total_goals = len(data[data['event'] == 'Goal'])
    total_shots = len(data[data['event'] == 'Shot']) + total_goals
    return total_goals / total_shots * 100

shots_data['distance'] = get_distance(shots_data)
shots_data['is_far'] = (shots_data['distance'] > 50)
```

```python
long_throw_stats = shots_data.groupby('is_far').apply(calculate_conversion_rate).reset_index()
long_throw_stats.columns = ['long_throw', 'conversion_rate']
```

Это подтверждает хоккейную интуицию: бросок от синей линии — это скорее попытка создать отскок или трафик перед воротами, чем реальная угроза голу.
```python
power_play_game_data['matchTime'] = (power_play_game_data['period'] - 1) * 1200 + power_play_game_data['periodTime']
```

```python
def get_powerplay_time(goal_row, data):
    # Ищем удаления в текущем или предыдущем периоде
    relevant_penalties = data[
        (data['game_id'] == goal_row['game_id']) &
        (data['matchTime'] <= goal_row['matchTime'])
    ]

    if not relevant_penalties.empty:
        last_penalty = relevant_penalties.iloc[-1]
        return goal_row['matchTime'] - last_penalty['matchTime']
    return None
```

### Гипотеза 10 : Волевые победы редки: Команда, проигрывающая после 2-го периода, имеет менее 20% шансов на итоговую победу в основное время

```python
# game_plays_data = pd.read_csv('/content/drive/MyDrive/archive/game_plays.csv')
```

```python
period2_scores = (
    game_plays_data[game_plays_data['period'] == 2]
    .groupby('game_id')
    .agg({'goals_home': 'last', 'goals_away': 'last'})
    .reset_index()
)

final_scores = (
    game_plays_data
    .groupby('game_id')
    .agg({'goals_home': 'last', 'goals_away': 'last'})
    .reset_index()
    .rename(columns={'goals_home': 'final_home', 'goals_away': 'final_away'})
)

scores = period2_scores.merge(final_scores, on='game_id')

scores['leader_after_2'] = scores.apply(
    lambda r: 'home' if r['goals_home'] > r['goals_away']
         else 'away' if r['goals_away'] > r['goals_home']
         else 'draw', axis=1
)

scores['final_winner'] = scores.apply(
    lambda r: 'home' if r['final_home'] > r['final_away']
         else 'away' if r['final_away'] > r['final_home']
         else 'draw', axis=1
)

result = scores[scores['leader_after_2'] != 'draw']

result['comeback'] = (
    (result['leader_after_2'] == 'home') & (result['final_winner'] == 'away') |
    (result['leader_after_2'] == 'away') & (result['final_winner'] == 'home')
)
```

```python
outcomes = result['comeback'].value_counts()
labels = ['Нет камбэка', 'Камбэк']
colors = ['#4C72B0', '#C44E52']

comeback_rate = result['comeback'].mean() * 100

plt.figure(figsize=(7, 7))
plt.pie(outcomes, labels=labels, colors=colors, autopct='%1.1f%%')
plt.title(f'Исходы матчей, где команда проигрывала после 2-го периода \n (камбэков: {comeback_rate:.1f}%)')
plt.show()
```

**Вывод:** Гипотеза **полностью подтвердилась**, причём с большим запасом. Реальная доля камбэков (10.7%) оказалась почти вдвое ниже заявленного порога в 20%. Это очень сильная закономерность: если команда проигрывает перед третьим периодом, её шансы на спасение крайне малы. С практической точки зрения это означает, что счёт после 2-го периода — один из самых надёжных предикторов итогового результата. Третий период редко меняет судьбу матча, и хоккейная драма с волевой победой — это действительно исключение, а не правило

### Гипотеза 11 : Первый гол решает: Команда, забившая первой, выигрывает матч более чем в 65% случаев

```python
first_goals = (
    game_plays_data[game_plays_data['event'] == 'Goal']
    .sort_values(['game_id', 'period', 'periodTime'])
    .groupby('game_id')
    .first()
    .reset_index()[['game_id', 'team_id_for', 'goals_home', 'goals_away']]
)
first_goals['first_scorer'] = first_goals.apply(
    lambda r: 'home' if (r['goals_home'] > 0 and r['goals_away'] == 0 )
         else 'away', axis=1
)

final_scores = (
    game_plays_data
    .groupby('game_id')
    .agg({'goals_home': 'last', 'goals_away': 'last'})
    .reset_index()
)

first_and_final = first_goals.merge(final_scores, on='game_id', suffixes=('_first', '_final'))

first_and_final['winner'] = first_and_final.apply(lambda r: 'home' if r['goals_home_final'] > r['goals_away_final'] else 'away', axis=1)

first_and_final['first_goal_won'] = (
    first_and_final['first_scorer'] == first_and_final['winner']
)
```

```python
outcomes = first_and_final['first_goal_won'].value_counts()
labels = ['Выиграл', 'Проиграл']
colors = ['#4C72B0', '#C44E52']

winate = first_and_final['first_goal_won'].mean() * 100

plt.figure(figsize=(6, 6))
plt.pie(outcomes, labels=labels, colors=colors, autopct='%1.1f%%')
plt.title(f'Исход матча для команды, забившей первой\n(побед: {winate:.1f}%)')
plt.show()
```

**Вывод:** Гипотеза **полностью подтвердилась**. Результат (67.9%) превышает заявленный порог в 65%, хоть и с небольшим запасом. Это подтверждает важное хоккейное правило: первый гол даёт не только психологическое преимущество, но и серьёзное тактическое — забившая команда может играть более закрыто и контратаковать, вынуждая соперника рисковать. Практическая ценность вывода: признак «кто забил первый гол» является мощным предиктором для модели прогнозирования исхода матча и должен быть включён в финальную модель наряду с фактором домашней площадки

```python
# game_plays_data = pd.read_csv('/content/drive/MyDrive/archive/game_plays.csv')
# game_plays_players_data = pd.read_csv('/content/drive/MyDrive/archive/game_plays_players.csv')
# players_info_data = pd.read_csv('/content/drive/MyDrive/archive/player_info.csv')
game_sketres_data = pd.read_csv('/content/drive/MyDrive/archive/game_skater_stats.csv')
```

```python
ice_time = game_sketres_data[['game_id', 'player_id', 'timeOnIce']].copy()
ice_time['rank'] = ice_time.groupby('game_id')['timeOnIce'].rank(ascending=False)

# делим на топ-3 (лидеры) и остальные (ранг 10+)
leaders = ice_time[ice_time['rank'] <= 3].copy()
bottom = ice_time[ice_time['rank'] >= 10].copy()
leaders['group'] = 'Топ-3 (лидеры)'
bottom['group'] = '3-4 звено'

penalties = game_plays_data[game_plays_data['event'] == 'Penalty'][['play_id', 'game_id']]
penalty_players = penalties.merge(
    game_plays_players_data[['play_id', 'player_id']], on='play_id'
)

penalty_counts = (
    penalty_players
    .groupby(['game_id', 'player_id'])
    .size()
    .reset_index(name='penalty_count')
)

leaders = leaders.merge(penalty_counts, on=['game_id', 'player_id'], how='left').fillna(0)
bottom = bottom.merge(penalty_counts, on=['game_id', 'player_id'], how='left').fillna(0)

all_players = pd.concat([leaders, bottom])
```

```python
means = all_players.groupby('group')['penalty_count'].mean().reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(data=means, x='group', y='penalty_count',
            palette={'Топ-3 (лидеры)': '#4C72B0', '3-4 звено': '#C44E52'})
plt.title('Среднее количество удалений за матч: лидеры vs игроки 3-4 звена')
plt.xlabel('')
plt.ylabel('Среднее количество удалений')

for i, val in enumerate(means['penalty_count']):
    plt.text(i, val + 0.002, f'{val:.4f}', ha='center', fontsize=12)

plt.show()
```

**Вывод:** Гипотеза **полностью подтвердилась**. Лидеры команды действительно фолят значимо реже, чем игроки третьего-четвертого звеньев. Это объясняется сразу несколькими факторами: во-первых, топ-игроки обладают лучшим катанием и позиционным чутьём, что позволяет им реже прибегать к фолам как к последнему средству; во-вторых, тренеры сознательно требуют от лидеров дисциплины, так как удаление звезды — двойная потеря (команда лишается лучшего игрока на 2 минуты); в-третьих, игроки нижних звеньев часто выполняют роль «чекеров» и вынуждены действовать агрессивнее и на грани фола

### Гипотеза 15 : Европейцы против североамериканцев: Европейские защитники делают в среднем меньше силовых приемов (hits), но больше перехватов, чем североамериканские

```python
# game_plays_data = pd.read_csv('/content/drive/MyDrive/archive/game_plays.csv')
# game_plays_players_data = pd.read_csv('/content/drive/MyDrive/archive/game_plays_players.csv')
players_info_data = pd.read_csv('/content/drive/MyDrive/archive/player_info.csv')
# game_sketres_data = pd.read_csv('/content/drive/MyDrive/archive/game_skater_stats.csv')
```

```python
defensemen_info = players_info_data[players_info_data['primaryPosition'] == 'D'].copy()

europe = ['RUS', 'SWE', 'FIN', 'CZE', 'SVK', 'SUI', 'DEU', 'LVA', 'BLR', 'DNK',
          'NOR', 'AUT', 'SVN', 'UKR', 'POL', 'LTU', 'EST', 'HRV', 'SRB', 'GBR', 'FRA']
north_america = ['CAN', 'USA']

defensemen_info['region'] = defensemen_info['nationality'].apply(lambda x: 'Европа' if x in europe else 'Северная Америка' if x in north_america else 'Другое')
defensemen_info = defensemen_info[defensemen_info['region'].isin(['Европа', 'Северная Америка'])]

hits = game_plays_data[game_plays_data['event'] == 'Hit'][['play_id', 'game_id']]
hits_players = hits.merge(game_plays_players_data[['play_id', 'player_id']], on='play_id')
hits_count = hits_players.groupby(['game_id', 'player_id']).size().reset_index(name='hits')

takeaways = game_plays_data[game_plays_data['event'] == 'Takeaway'][['play_id', 'game_id']]
takeaways_players = takeaways.merge(game_plays_players_data[['play_id', 'player_id']], on='play_id')
takeaways_count = takeaways_players.groupby(['game_id', 'player_id']).size().reset_index(name='takeaways')

def_stats = defensemen_info[['player_id', 'region']]
def_stats = def_stats.merge(hits_count, on='player_id', how='left')
def_stats = def_stats.merge(takeaways_count, on=['game_id', 'player_id'], how='left')
def_stats[['hits', 'takeaways']] = def_stats[['hits', 'takeaways']].fillna(0)

region_means = def_stats.groupby('region')[['hits', 'takeaways']].mean().reset_index()

region_melt = region_means.melt(id_vars='region', var_name='metric', value_name='avg_per_game')
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(data=region_melt, x='metric', y='avg_per_game', hue='region',
            palette={'Европа': '#4C72B0', 'Северная Америка': '#C44E52'})
plt.title('Защитники: Европа VA Северная Америка')
plt.ylabel('Среднее количество за матч')

for i, bar in enumerate(plt.gca().patches):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.002,
             f'{bar.get_height():.3f}', ha='center', fontsize=10)

plt.show()
```

**Вывод:** Гипотеза **полностью опровергнута**. Вопреки стереотипу о «мягких» европейских и «жёстких» североамериканских защитниках, данные не показывают значимой разницы ни в силовых приёмах, ни в перехватах. Более того, европейцы даже немного чаще проводят хиты (5.48 против 5.18), а по перехватам разница и вовсе статистически незначима (0.49 против 0.52). Это интересное опровержение популярного хоккейного мифа: на уровне НХЛ стиль игры защитника определяется не национальностью, а индивидуальными качествами, системой команды и тренерскими установками. Национальные стереотипы о «европейском» и «североамериканском» стиле, по-видимому, не находят отражения в реальной статистике
